## Data Cleaning

### Filtering aircrafts and events

We want to filter the dataset to include aircraft that the client is interested in an analysis of:

```python
# Make a copy
airline_df = df.copy(deep=True)
```

Observation:Column names have been changed to lowercase with underscores replacing dots, making them easier to work with.

```python
# Check unique values in aircraft_category
airline_df['aircraft_category'].value_counts()
```

```python
# Filter Airplanes 
airline_df = airline_df[airline_df['aircraft_category'] == 'Airplane']

# Check
airline_df.shape
```

```python
# Check unique values in amateur_built
airline_df['amateur_built'].value_counts()
```

```python
# Check event_date range
airline_df['event_date'].value_counts()
```

```python
# Filter for safe flights
airline_df = airline_df[airline_df['amateur_built'] == 'No']
airline_df.shape
```

```python
# Convert event_date to datetime
airline_df['event_date'] = pd.to_datetime(airline_df['event_date'])
```

```python
# Filter flights from 1983
airline_df = airline_df[airline_df['event_date'].dt.year >= 1983]
airline_df.shape
```

### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
# Check injury columns
airline_df[['total_fatal_injuries', 'total_serious_injuries', 'total_minor_injuries', 'total_uninjured']].head(10)
```

```python
# Fill NaN values with 0 for injury columns
airline_df['total_fatal_injuries'] = airline_df['total_fatal_injuries'].fillna(0)
airline_df['total_serious_injuries'] = airline_df['total_serious_injuries'].fillna(0)
airline_df['total_minor_injuries'] = airline_df['total_minor_injuries'].fillna(0)
airline_df['total_uninjured'] = airline_df['total_uninjured'].fillna(0)
airline_df
```

Observation:Missing values in injury columns have been filled with 0, assuming no injuries were recorded.

```python
# number of passengers on each flight
airline_df['total_passengers'] = (airline_df['total_fatal_injuries'] + 
                                   airline_df['total_serious_injuries'] + 
                                   airline_df['total_minor_injuries'] + 
                                   airline_df['total_uninjured'])
airline_df['total_passengers'].head()
```

```python
# Calculate fraction of passengers seriously/fatally injured
airline_df['injury_fraction'] = (airline_df['total_fatal_injuries'] + 
                                  airline_df['total_serious_injuries']) / airline_df['total_passengers']
```

```python
# Check the new column
airline_df['injury_fraction'].describe()
```

Observation:The mean injury fraction is 0.28, meaning on average 28% of passengers are seriously or fatally injured in an accident.

**Aircraft.Damage**

- construct a derived column tracking whether an aircraft was destroyed or not.

```python
# Check unique values in aircraft_damage
airline_df['aircraft_damage'].value_counts()
```

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

```python
# Check unique values in make column
airline_df['make'].value_counts()
```

```python
# Capitalize the first letter of each
airline_df['make'] = airline_df['make'].str.strip().str.title()

# Check it
airline_df['make'].value_counts()
```

Observation:duplicate entries like CESSNA and Cessna have been merged.
